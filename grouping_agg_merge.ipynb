# Grouping, Aggregation and Merging
- will learn about groupby and other methods
- aggregate function like sum, min, max, avg and more
- merging similar to sql table joins

```python
import pandas as pd

matches = pd.read_csv('data/matches.csv')
deliveries = pd.read_csv('data/deliveries.csv')
```

```python
# Find no of matches played in each season
matches.groupby(by="season")["season"].size().sort_values(ascending=False)

#How many matches eack team won
matches.groupby("winner")["winner"].count()

#total run scored by each batting team
deliveries.groupby("batting_team")["total_runs"].sum()
```

```python

batsman_group = deliveries.groupby('batsman')

def batsman_total_runs():
    return batsman_group['batsman_runs'].sum().sort_values(ascending=False)

# For each batter find total runs, max run in a ball, avg run per ball
def batsman_stat():
   return batsman_group.agg(
        max_run_in_ball= ('batsman_runs', 'max'),
        avg_runs =  ('batsman_runs', 'mean'),
        total_runs = ('batsman_runs', 'sum')
    ).sort_values(by=['total_runs'],ascending=False)
   
#  Find runs scored by each batsman against each bowling team
def batsman_record_against_teams():
    return deliveries.groupby(by=["batsman", "bowling_team"])['batsman_runs'].sum().sort_values(ascending=False).head(10)

#Calculate teams avg runs for every delivery 
def calculate_team_avg_runs():
    team_group = deliveries.groupby("batting_team")
    deliveries['team_avg_run'] = team_group['total_runs'].transform('mean')
    return deliveries.groupby('batting_team')['team_avg_run'].first().sort_values(ascending=False)
    
# batsman_stat()
# batsman_record_against_teams()
calculate_team_avg_runs()
```
